```python
x=df.drop(columns=["price"],axis=1)
y=df["price"]
```

```python
x
```

```python
x_one_encoded=pd.get_dummies(x,drop_first=True)
```

```python
x_one_encoded
```

```python
x_one_encoded=x_one_encoded.astype(int)
x_one_encoded
```

```python
x_label=x.copy()
x_label
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
x_label["model"]=le.fit_transform(x_label["model"])
x_label["transmission"]=le.fit_transform(x_label["transmission"])
x_label["fuelType"]=le.fit_transform(x_label["fuelType"])
```

```python
x_label
```

```python
numeric_col=["year","mileage","tax","mpg","engineSize"]
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_one_encoded[numeric_col] = sc.fit_transform(x_one_encoded[numeric_col])
x_one_encoded
```

```python
x_label[numeric_col] = sc.fit_transform(x_label[numeric_col])
x_label
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error , mean_squared_error,r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x_one_encoded, y, test_size=0.33, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1))/(n-p-1)
print("Adjusted R2 score: ",adjusted_r2)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x_label, y, test_size=0.33, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
r2=r2_score(y_test,y_pred)
r2
```
